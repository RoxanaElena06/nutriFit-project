```python
df = (spark.read.format("csv").option("header", "true").option("inferSchema", "true").load("/Volumes/nutrifit_catalog/bronze/raw_files/daily_activity.csv"))
display(df.limit(5))
df.printSchema()
```

```python
display(df.select("id", "ActivityDate", "TotalSteps", "Calories"))
```

```python
# check for negative calories
df.filter(df.Calories < 0).count()
```

```python
# check duplicates
total = df.count()
unique = df.dropDuplicates().count()
print(f"Total rows: {total}, unique rows: {unique} and duplicate rows: {total - unique}")
```
